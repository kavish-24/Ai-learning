```python
import chromadb

client = chromadb.PersistentClient(path="./chroma_db")

collection = client.get_or_create_collection(
    name="employee_documents"
)

print(collection.name)
```

```python
collection.add(
    ids=["1", "2", "3"],
    documents=[
        "Python is a programming language.",
        "RAG stands for Retrieval Augmented Generation.",
        "Embeddings convert text into numerical vectors."
    ]
)
```

```python
collection.add(
    ids=["4", "5"],
    documents=[
        "Employees receive 12 casual leaves annually.",
        "Employees receive 15 vacation leaves annually."
    ],
    metadatas=[
        {"page": 2, "source": "policy.pdf"},
        {"page": 3, "source": "policy.pdf"}
    ]
)
```

```python
ans = collection.query(
    query_texts=["What is machine learning?"],
    n_results=2
)
```

```python
import faiss
import numpy as np
from sentence_transformers import SentenceTransformer
model = SentenceTransformer('all-MiniLM-L6-v2')
documents = [
    "Python is a programming language.",
    "RAG retrieves external information.",
    "Dogs are domestic animals."
]

embeddings = model.encode(documents)

embeddings = np.array(embeddings).astype("float32")

#create a FAISS index

dimension = embeddings.shape[1]
index = faiss.IndexFlatL2(dimension)

index.add(embeddings)

# Search
query = "What is RAG?"

query_embedding = model.encode([query])
query_embedding = np.array(query_embedding).astype("float32")

distances, indices = index.search(query_embedding, 2)

for i in indices[0]:
    print(documents[i])
```
