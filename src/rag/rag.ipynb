```python
paragraph = """
Employees are entitled to 12 casual leaves every year.
Casual leaves can be taken for personal work or emergencies.
Employees must inform their managers before taking leave.
Unused casual leaves cannot be carried forward to the next year.
Sick leaves are provided separately, and employees must submit
medical certificates when taking more than three consecutive days off.
The company also provides 15 days of paid vacation annually.
Employees should submit vacation requests at least one week in advance.
"""

word = paragraph.split()
chunk_size = 30
overlap = 5

chunks = []

for i in range(0 , len(word) , chunk_size - overlap):
    chunk = word[i:i + chunk_size]
    chunks.append(" ".join(chunk))
    
    
for index, chunk in enumerate(chunks):
    print(f"Chunk {index + 1}: {chunk}\n")
```

```python
import pymupdf
pdf = pymupdf.open(r"D:\langchain\src\rag\SDE_Interview_Theory_Notes.pdf")
documents = []
```

```python
for page_number ,page in enumerate(pdf):
    text = page.get_text()
    
    documents.append({
      "text": text,
      "page": page_number+1,
      "source": "D:\langchain\src\rag\SDE_Interview_Theory_Notes.pdf"
    })
    
print(documents)
```

```python
chunks = []

for document in documents:

    words = document["text"].split()

    chunk_size = 100
    overlap = 20

    for i in range(0, len(words), chunk_size - overlap):

        chunk_text = " ".join(words[i:i + chunk_size])

        chunks.append({
            "text": chunk_text,
            "page": document["page"],
            "source": document["source"]
        })
```

```python
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity
from sentence_transformers import SentenceTransformer
model = SentenceTransformer('all-MiniLM-L6-v2')
texts = [chunk["text"] for chunk in chunks]

document_embeddings = model.encode(texts)

def retrieve(query):

    query_embedding = model.encode([query])

    scores = cosine_similarity(
        query_embedding,
        document_embeddings
    )[0]

    top_indices = np.argsort(scores)[::-1][:3]

    results = []

    for index in top_indices:
        result = chunks[index]

        results.append({
            "text": result["text"],
            "page": result["page"],
            "source": result["source"],
            "score": float(scores[index])
        })

    return results
```

```python
from typing import TypedDict

class Chunk(TypedDict):
    text: str
    page: int
    source: str

class RetrievedChunk(Chunk):
    score: float
```

```python
def retrieve(query: str, k: int = 3) -> list[RetrievedChunk]:
    query_embedding = model.encode([query])
    scores = cosine_similarity(query_embedding, document_embeddings)[0]
    top_indices = np.argsort(scores)[::-1][:k]

    return [
        RetrievedChunk(
            text=chunks[i]["text"],
            page=chunks[i]["page"],
            source=chunks[i]["source"],
            score=float(scores[i]),
        )
        for i in top_indices
    ]
```

```python
from dotenv import load_dotenv
from langchain.chat_models import init_chat_model

load_dotenv()

# `model` is the embedding model, so the chat LLM gets its own name
llm = init_chat_model("groq:openai/gpt-oss-20b")
```

```python
from langchain.tools import tool

@tool
def search_documents(query: str) -> str:
    """Search the SDE interview notes PDF. Returns relevant passages with their page number and source file."""
    results = retrieve(query)
    return "\n\n".join(
        f"[Source: {r['source']} | Page: {r['page']} | Score: {r['score']:.2f}]\n{r['text']}"
        for r in results
    )
```

```python
from langchain.agents import create_agent

agent = create_agent(
    llm,
    tools=[search_documents],
    system_prompt=(
        "You answer questions using the search_documents tool. "
        "Always call it first, answer only from the results, "
        "and cite the page number and source file for every claim."
    ),
)
```

```python
response = agent.invoke(
    {"messages": [{"role": "user", "content": "What is the difference between a process and a thread?"}]}
)
print(response["messages"][-1].content)
```
