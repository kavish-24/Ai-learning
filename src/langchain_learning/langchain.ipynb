```python
import os
from  dotenv import load_dotenv
load_dotenv()

os.environ["GROQ_API_KEY"]=os.getenv("GROQ_API_KEY")
print("success")


```

```python
from langchain.agents import create_agent
def get_wheather(city:str)->str:
    """returns whether for city"""
    return f"the whether for {city} is sunny"

agent= create_agent(
    model="groq:openai/gpt-oss-20b",
    tools=[get_wheather],
    system_prompt="You are a helpful assitent."

)
agent

```

```python
###RUN THE AGENT
response= agent.invoke({"messages":[{"role":"user","content":"whats the whether in goa?"}]})
```
