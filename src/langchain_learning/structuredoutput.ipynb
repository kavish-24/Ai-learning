```python
import os 
from dotenv import load_dotenv
from langchain.chat_models import init_chat_model
load_dotenv()
os.environ["api_key"]=os.getenv("GROQ_API_KEY")

model = init_chat_model("groq:openai/gpt-oss-20b")
```

```python
from pydantic import BaseModel ,  Field

class Movie(BaseModel):
    title:str =Field(...,description="title of the movie")
    year:int= Field(...,description="YEar of movie release")
    director:str=Field(...,description="movie director")
    rating:float= Field(...,description="movie rating out of 10")
```

```python
model_with_structured_output=model.with_structured_output(Movie , include_raw=True)
model_with_structured_output
```

```python
model_with_structured_output.invoke("provide me details of movie inception")
```

```python
#nested structure

class Actor(BaseModel):
    name:str
    role:str
    
    
class MovieDetail(BaseModel):
    title:str
    year:int
    cast : list[Actor]
    genre:list[str]
    budget:float|None=Field(None, description="Budget in millions crores")    
```

```python
model_with_structured_output=model.with_structured_output(MovieDetail)
model_with_structured_output.invoke("Provide me details on Bahubali")
```

Typedict

```python
from typing_extensions import TypedDict , Annotated
```

```python
class Movie(TypedDict):
    """A movie with details"""
    title:Annotated[str,...,"title of the movie"]
    year: Annotated[int,...,"YEar of movie release"]
    director:Annotated[str,...,"movie director"]
    rating: Annotated[float,...,"movie rating out of 10"]
```

```python
mode_with_typedist=model.with_structured_output(Movie)
```

```python
response= mode_with_typedist.invoke("provide me details on movie avengers")
response
```

```python
from pydantic import BaseModel

class User(BaseModel):
    id: int
    name: str
    email: str
    age: int

# Creating an object
user = User(
    id=1,
    name="Kavish",
    email="kavish@gmail.com",
    age=22
)

print(user.name)
print(user.model_dump())
```

```python
from dataclasses import dataclass
from langchain.agents import create_agent


@dataclass
class ContectInfo:
    """Contectinfo for a person"""
    name:str
    email:str
    phone:str
    
    
model = create_agent(
    model= "gpt-oss-20b",
    response_format=ContectInfo
    
)
```
