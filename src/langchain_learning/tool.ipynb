```python
import os
from dotenv import load_dotenv

load_dotenv()

os.environ["groq_api_key"] = os.getenv("GROq_API_KEY")
os.environ["gemini_api_key"] = os.getenv("GEMINI_API_KEY")
```

```python
#### google gemini model 

from langchain_groq import ChatGroq

model = ChatGroq(model="openai/gpt-oss-20b")
response = model.invoke("who r u?")
response.content
```

```python
from langchain.tools import tool


@tool
def get_weather(location:str)->str:
   """Get weather for the location"""
   return f"its sunny in {location}"

model_with_tools= model.bind_tools([get_weather])
```

```python
response = model_with_tools.invoke("what is the weather in boston?")

print(response) 
```

```python
print(response.content)
```

```python
for tool_call in response.tool_calls:
    print(f"TOOLs:{tool_call['name']}")
    print(f"arg:{tool_call['args']}")
```

```python
from langchain_core.messages import HumanMessage

messages = [HumanMessage("what is the weather in boston?"), response]
for tool_call in response.tool_calls:
    messages.append(get_weather.invoke(tool_call))

final = model_with_tools.invoke(messages)
print(final.content)
```
