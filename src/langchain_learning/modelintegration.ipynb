```python
#### google gemini model 

from langchain.chat_models import init_chat_model
model = init_chat_model("google_genai:gemini-2.5-flash-lite")
response = model.invoke("who r u?")
response.content
```

```python
#### google gemini model 

from langchain_groq import ChatGroq

model = ChatGroq(model="openai/gpt-oss-20b")
response = model.invoke("who r u?")
response.content
```

Streaming and batch

```python
for chunk in model.stream("write me a 200 words paragragh on Ai"):
    print(chunk.text, end="", flush=True)
```

batch

```python
reponses = model.batch([
    "what is color?",
    "who invented gravity",
    "who is modi"
],
                       config={
                           "max_concurrency":5,
                       })

```

```python
for result in reponses:
    print(result)
```
